1. What do each of the sensors measure physiologically? [10]

Inertial Measurement Units (IMUs) measure linear acceleration and angular velocity in three dimensions (x, y, z). When placed on the thighs and back, they capture movement of the lower limbs and trunk, providing information about hip and trunk motion. These signals correspond to anatomical movements in the anterior–posterior, medial–lateral, and superior–inferior directions, allowing physiological interpretation of gait, posture, and overall body movement.

Galvanic Skin Sensor (GSS) - Applies a small current and assesses the skin's resistance between two electrodes, resultantly measuring electrodermal activity (EDA) which is the skin's electrical properties due to emotional arousal and stress. https://www.sciencedirect.com/topics/computer-science/galvanic-skin-response 

Electroencephalography (ECG) - This sensor measures electrical activity of the heart (heart beat, irregular heart beats, arrhythmias, and heart attacks) using electrodes that are placed on the left arm, right arm, and right leg based on Einthoven’s triangle. The output of this sensor is a single voltage signal that combines information from the three electrodes [reference the original assignment]. https://www.mayoclinic.org/tests-procedures/ekg/about/pac-20384983 




2. How do you expect the physiology that is measured change when a fall or near-fall occurs
compared to when participants are experiencing an activity of daily living? (e.g. why are these
sensors potentially useful for the grand challenge?) [15]


During falls and near-falls, physiological signals are expected to differ significantly from those seen during activities of daily living (ADLs) because these events involve sudden instability and stress. IMU data would show abrupt spikes in linear acceleration and angular velocity due to loss of balance and rapid corrective movements, whereas ADLs typically produce smoother, more predictable motion patterns. This makes IMUs particularly useful for identifying biomechanical signatures of falls and near-falls.

Cardiovascular and autonomic responses are also expected to change. ECG signals may show rapid increases in heart rate during falls or near-falls due to activation of the sympathetic nervous system, while galvanic skin conductance may increase as a result of stress or startle responses. In contrast, ADLs generally produce more gradual or stable physiological changes unless they involve intense physical exertion.
These sensors are useful for the grand challenge because they capture complementary information: IMUs detect physical instability, while ECG and GSS reflect physiological stress. Together, they provide a multidimensional view of falls and near-falls, improving the ability to distinguish them from normal daily activities.
